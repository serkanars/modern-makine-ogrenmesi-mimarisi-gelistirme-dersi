## 9. `for` döngüsü: koleksiyon üzerinde ilerleme

Aynı işlemi listenin her elemanı için tekrar etmek istediğimizde `for` kullanırız. `range`, `enumerate` ve `zip` sık kullanılan yardımcı yapılardır.

````python
öğrenciler = ["a","b","c"]

for öğrenci in öğrenciler:
    print(öğrenci)
````

````python
customers = ["Ayşe", "Mehmet", "Zeynep"]
fees = [450, 320, 510]

for customer_name in customers:
    print("Müşteri:", customer_name)

for index, customer_name in enumerate(customers, start=1):
    print(index, customer_name)

for customer_name, fee in zip(customers, fees):
    print(f"{customer_name}: {fee} TL")
````

## 12. Fonksiyonlar: tekrar kullanılabilir kod

Fonksiyon belirli bir işi isimlendirir. Girdilere **parametre**, fonksiyonu çağırırken verdiğimiz değerlere **argüman** deriz. `return`, sonucu fonksiyonun dışına verir. `print` yalnızca ekrana yazar; `return` edilen değer başka bir hesapta kullanılabilir.

````python
def benim_adim(ad, soyad, yaş):
    print(f"benim adım : {ad}")
    return ad
````

````python
benim_adim("serkan")
````

````python
def calculate_discounted_fee(monthly_fee, discount_rate=0.10):
    """Aylık ücrete indirim uygular ve yeni ücreti döndürür."""
    if monthly_fee < 0:
        raise ValueError("monthly_fee negatif olamaz")
    if not 0 <= discount_rate <= 1:
        raise ValueError("discount_rate 0 ile 1 arasında olmalıdır")
    return monthly_fee * (1 - discount_rate)

new_fee = calculate_discounted_fee(450, 0.15)
print("Yeni ücret:", new_fee)
````

### Scope: değişken nerede geçerli?

Fonksiyon içinde oluşturulan değişkenler genellikle yalnızca fonksiyon içinde yaşar. Fonksiyonun dışındaki global değişkenleri değiştirmek yerine, gerekli değerleri parametre olarak alıp sonucu `return` etmek daha güvenlidir.

````python
tax_rate = 0.20  # global

def add_tax(price, rate):
    final_price = price * (1 + rate)  # local
    return final_price

print(add_tax(100, tax_rate))
# print(final_price)  # Açarsak NameError alırız; değişken fonksiyonun dışında yoktur.
````

# Mini Proje — Müşteri risk kural motoru

Şimdi öğrendiğimiz dictionary, list, koşul, döngü ve fonksiyonları bir araya getireceğiz. Bu bir makine öğrenmesi modeli değildir. Kuralları biz yazıyoruz; sistem veriden öğrenmiyor. İlerleyen haftalarda aynı problemi öğrenen modellere taşıyacağız.

Kuralımız:
- Destek çağrısı 4 veya üzeriyse `+2` puan
- Kullanım değişimi `%−30` veya daha düşükse `+2` puan
- Kontrat aylıksa `+1` puan
- Gecikmiş ödeme varsa `+1` puan
- Toplam 4 veya üzeriyse yüksek, 2–3 ise orta, aksi halde düşük risk


````python
customer_records = [
    {"customer_id": "C001", "support_calls": 5, "usage_change_pct": -42, "contract": "monthly", "late_payment": True},
    {"customer_id": "C002", "support_calls": 1, "usage_change_pct": 5, "contract": "annual", "late_payment": False},
    {"customer_id": "C003", "support_calls": 3, "usage_change_pct": -18, "contract": "monthly", "late_payment": False},
]

def calculate_risk(customer):
    score = 0
    reasons = []

    if customer["support_calls"] >= 4:
        score += 2
        reasons.append("yüksek destek çağrısı")

    if customer["usage_change_pct"] <= -30:
        score += 2
        reasons.append("kullanım düşüşü")

    if customer["contract"] == "monthly":
        score += 1
        reasons.append("aylık kontrat")

    if customer["late_payment"]:
        score += 1
        reasons.append("gecikmiş ödeme")

    if score >= 4:
        level = "yüksek"
    elif score >= 2:
        level = "orta"
    else:
        level = "düşük"

    return {"score": score, "level": level, "reasons": reasons}

for customer in customer_records:
    result = calculate_risk(customer)
    print(customer["customer_id"], "→", result)
````
