## List Compheresion

```python
def  topla_ve_böl(sayi_1: int, sayi_2: float):
    "Bu fonksiyon verilen iki sayıyı toplar ve böler."
    
    if type(sayi_1) != int:
        raise ValueError("ilk sayı her zaman integer-doğal sayı olmalıdır.")
    
    return (sayi_1 + sayı_2), (sayi_1 / sayı_2)
```

```python
topla_ve_böl(sayi_1=8.5, sayi_2=19)
```
